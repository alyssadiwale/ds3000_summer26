## Part 2.2: Exploring Arrays (15 points)

1. Give the shape, size, ndim, and nbytes for each of the two arrays.
1. Take the transpose of both arrays. Call these `t_array_a` and `t_array_b`.
1. Try to add `array_a` and `t_array_b` (*prove* and *show* you did this with commented out code), then remove the last column of `t_array_b` and try to add them again. In a markdown cell, explain what happened.

```python
for name, arr in [("array_a", array_a), ("array_b", array_b)]:
    print(f"{name}:")
    print(f"  shape:  {arr.shape}")
    print(f"  size:   {arr.size}")
    print(f"  ndim:   {arr.ndim}")
    print(f"  nbytes: {arr.nbytes}")
    print()

t_array_a = array_a.T
t_array_b = array_b.T

print(t_array_a)
print(t_array_b)

# Attempting to add array_a and t_array_b — this will FAIL
# array_a is (3,4) and t_array_b is (3,5) — shapes don't match
# array_a + t_array_b  # <-- uncomment to see the ValueError

# Remove the last column of t_array_b to make it (3,4)
t_array_b_trimmed = t_array_b[:, :-1]

# Now both are (3,4) — this works!
result = array_a + t_array_b_trimmed
print(result)

#The first addition failed with a `ValueError` because `array_a` has shape **(3, 4)** and `t_array_b` has shape **(3, 5)** — NumPy requires arrays to have matching shapes  (or compatible shapes for broadcasting) to perform element-wise addition.

# By removing the last column of `t_array_b` with `[:, :-1]`, we reduced it to shape **(3, 4)**, matching `array_a`, so the addition succeeded.
```

## Part 3.2: Manipulating Data (20 points)

1. Create a subset of the data set which (a) **includes** only train stations in Belgium and (b) **excludes** all train stations which are **not** in a city. Make sure to save this subset as a new data frame and print the first few rows of the data.
1. Use the `.describe()` function to produce summary statistics for the subset from the previous part. Create a markdown cell and explain:
    - What Series did the `.describe()` function run on? What Series did it not run on? What is the difference, and what does this mean the `.describe()` function is used for?

```python
# (a) only Belgium, (b) exclude stations not in a city (drop rows where city is NaN)

import pandas as pd

df = pd.read_csv('train_stations_europe.csv', index_col='id')

# Part 3.1 - print first few rows
print(df.head())

# Part 3.2

# (a) only Belgium, (b) exclude stations not in a city (is_city == 1 means it is in a city)
belgium_df = df[(df['country'] == 'BE') & (df['is_city'] == 1)]

print(belgium_df.head())

# Summary statistics for belgium_df
print(belgium_df.describe())

#The `.describe()` function ran on the **numeric** columns only: `latitude`, `longitude`, `is_city`, `is_main_station`, and `is_airport`.

#It did **not** run on **string/object** columns like `name`, `country`, and `time_zone`, because operations like mean, std, and min/max are meaningless for text data.

#This means `.describe()` is used for **quickly summarizing the distribution of numerical data** — helpful for spotting outliers, checking value ranges, and understanding the overall shape of your numeric variables.
```
